# How Hindsight Became the Memory Layer in My Agent

I started this system with a fairly ordinary support-agent problem: the model could answer the current question, but it had no durable understanding of what had happened before.

I decided not to solve that by making the prompt larger. Instead, I made memory a separate layer and put Hindsight between the support agent and its past.

## The architecture I ended up with

The finished request path is deliberately simple:

```text
Customer
   |
   v
Frontend
   |
   v
Flask API
   |
   +---- Hindsight: recall relevant history
   |
   +---- Local LLM: generate the response
   |
   +---- Hindsight: retain the interaction
   |
   v
Frontend
```

The important part is the separation.

The agent does not store memories itself. The LLM does not own customer history. The frontend does not talk directly to Hindsight.

Instead, the backend asks a memory service for context and gives the resulting context to the model.

That made Hindsight a genuine infrastructure layer rather than another utility hidden inside the prompt-building code.

For the memory model and API, I worked from the [Hindsight GitHub repository](https://github.com/vectorize-io/hindsight) and [Hindsight documentation](https://hindsight.vectorize.io/).

## Why I didn't use chat history as memory

The obvious first implementation would have been to save every conversation and send previous messages back to the model.

That works for a small prototype.

It becomes less attractive when the history starts containing hundreds of messages.

More importantly, raw conversation history isn't necessarily the information I want the model to use.

A support interaction might contain:

```text
Customer reports a PDF upload crash
        ↓
Agent asks about browser
        ↓
Customer says Chrome on Windows 11
        ↓
Cache is cleared
        ↓
Problem remains
        ↓
Application is updated
        ↓
Problem is resolved
```

If the customer returns later and says:

```text
My application is crashing again.
```

I don't want to blindly replay the transcript.

I want the system to retrieve the pieces that are relevant to the new question.

That distinction is what pushed me toward a dedicated memory layer.

The broader idea is described well in the [agent memory overview from Vectorize](https://vectorize.io/what-is-agent-memory): durable agent memory is more than keeping a longer conversation context.

## The memory service became the boundary

I wanted the rest of the application to be unaware of Hindsight-specific details.

The agent calls:

```python
recall = self.memory.recall_customer_memory(
    customer_id=request.customer_id,
    query=request.message,
)
```

It doesn't know how the customer is mapped to storage, how retrieval works, or which Hindsight API method is being used.

That logic lives inside `MemoryService`.

This gave me a useful dependency direction:

```text
Agent
  |
  v
MemoryService
  |
  v
Hindsight client
  |
  v
Hindsight
```

If I later replace the memory provider, the agent shouldn't need to change.

That sounds like standard dependency inversion, but it matters more with an agent because the memory system tends to become entangled with prompting very quickly.

I wanted to avoid that.

## Recall is an explicit step

The core agent method makes the memory lifecycle visible:

```python
recall = self.memory.recall_customer_memory(
    customer_id=request.customer_id,
    query=request.message,
)

memory_available = bool(recall.success)

memories = (
    [self._memory_dict(item) for item in recall.memories]
    if recall.success
    else []
)
```

Then the retrieved memories are passed into the LLM request:

```python
response = self.llm.generate(
    instructions=SYSTEM_PROMPT,
    user_input=build_user_input(
        request.message,
        memories,
    ),
)
```

Only after that does the agent retain the new interaction.

That ordering matters:

```text
          CURRENT MESSAGE
                |
                v
        Hindsight RECALL
                |
                v
       Relevant historical
             context
                |
                v
          LLM RESPONSE
                |
                v
        Hindsight RETAIN
```

I don't want the new interaction influencing the memory retrieval for the same request. The model should answer using the state that existed before the current interaction, then the new state can be recorded.

## Hindsight isn't responsible for the answer

One design decision I kept coming back to was where reasoning should happen.

Hindsight retrieves memory.

The LLM generates the response.

The agent coordinates them.

That gives each component a narrow responsibility:

```text
Hindsight
  "What previous information is relevant?"

LLM
  "Given that information and the current request,
   what should I say?"

Agent
  "When should I recall, generate, and retain?"
```

This separation makes debugging much easier.

If the response ignores a relevant previous issue, I can inspect the recalled memories separately from the generated answer.

If nothing was recalled, I can investigate retrieval.

If the right memory was returned but the response ignored it, I can investigate the prompt or model behavior.

Without that separation, all three problems tend to look like "the AI gave a bad answer."

## I made the prompt explicitly memory-aware

Retrieval alone isn't enough.

The model has to understand what the retrieved information means.

My system prompt tells the model that the historical section contains real information from previous support interactions and gives it rules for using that information.

The input is structured as:

```python
return f"""
CURRENT CUSTOMER MESSAGE:
{message}

RELEVANT HISTORICAL MEMORIES:
{memory_text}

INSTRUCTIONS FOR THIS RESPONSE:

First determine whether any historical memory is relevant to the current
customer message.

If relevant history exists, use it in the response and naturally acknowledge
the previous issue. Do not ask the customer for information that the relevant
memory already provides.

If the current issue appears to be a continuation of a previous issue, clearly
connect the two.

Do not invent information.

Now provide the best next support response.
""".strip()
```

I deliberately distinguish between **retrieved** and **relevant**.

Just because Hindsight returned something doesn't mean the model should mention it.

That prevents the memory layer from becoming a dumping ground for old customer information.

## I also retain outcomes, not just messages

The retain side of the system is just as important as recall.

The agent records information such as:

```python
retain = self.memory.remember_interaction(
    customer_id=request.customer_id,
    name=request.name,
    problem=request.message,
    environment=request.environment,
    attempted_solutions=request.attempted_solutions,
    failed_solutions=request.failed_solutions,
    successful_solutions=request.successful_solutions,
    resolution=request.resolution,
    preferences=request.preferences,
    context=request.context,
    ticket_id=request.ticket_id,
)
```

That means the memory layer can represent more useful support history than a sequence of chat messages.

For example:

```text
Problem:
PDF upload crashes

Environment:
Windows 11 + Chrome

Attempt:
Clear application cache

Outcome:
Failed

Attempt:
Update application

Outcome:
Successful
```

When the customer returns with the same problem, that history can influence the next troubleshooting step.

The model is also instructed not to blindly repeat previously failed solutions.

That's one of the practical differences I wanted from persistent memory.

## Per-customer memory

I also needed a clear isolation boundary.

Each customer gets a separate Hindsight bank.

The service derives a deterministic bank identifier from the customer ID:

```python
def _bank_id(self, customer_id: str) -> str:
    digest = hashlib.sha256(
        customer_id.encode("utf-8")
    ).hexdigest()[:24]

    return f"memorycare-{digest}"
```

So:

```text
rahul_001  ->  one memory bank
priya_002  ->  another memory bank
arjun_003  ->  another memory bank
```

The agent doesn't have to remember to add a customer filter to every Hindsight query because the customer identity is part of the memory-service operation itself.

I consider that an important architectural property, not just an implementation detail.

## The Hindsight client forced me to think about async boundaries

The most painful integration issue wasn't retrieval quality or prompting.

It was execution context.

My Flask backend is synchronous, while the Hindsight client exposes asynchronous operations such as:

```text
acreate_bank
arecall
aretain
aget_version
```

My first direct integration produced:

```text
Timeout context manager should be used inside a task
```

That error initially looked unrelated to memory.

The actual issue was that the async HTTP client's lifecycle was being used from the wrong execution context.

Rather than spreading asyncio through the backend, I isolated it behind a synchronous adapter.

The adapter owns an event loop in a background thread:

```python
class HindsightSyncAdapter:
    def __init__(self, client):
        self._client = client
        self._loop = asyncio.new_event_loop()

        self._thread = threading.Thread(
            target=self._run_loop,
            name="hindsight-event-loop",
            daemon=True,
        )

        self._thread.start()

    def _run_loop(self):
        asyncio.set_event_loop(self._loop)
        self._loop.run_forever()
```

Calls are submitted safely to that loop:

```python
def _run(self, coroutine):
    future = asyncio.run_coroutine_threadsafe(
        coroutine,
        self._loop,
    )
    return future.result()
```

That kept the rest of the application synchronous.

It also gave me one place to reason about the Hindsight client's lifecycle.

## The LLM remains replaceable

I currently run Qwen3 locally through Ollama.

I didn't want that choice to become part of the memory architecture, so the agent depends on a small protocol:

```python
class LLMClient(Protocol):
    def generate(
        self,
        *,
        instructions: str,
        user_input: str
    ) -> str:
        ...
```

The concrete implementation can call the local Ollama API, but the memory service doesn't care.

That leaves the architecture looking like:

```text
             +----------------+
             |   Memory Layer |
             |   Hindsight    |
             +-------+--------+
                     |
               relevant context
                     |
                     v
+----------+    +---------+    +----------+
| Frontend | -> |  Agent  | -> |   LLM    |
+----------+    +---------+    +----------+
                     |
                     v
                new memory
```

If I change the LLM provider, Hindsight stays where it is.

If I change the memory provider, the agent's interface can remain the same.

That independence was intentional.

## What the system looks like in practice

The useful behavior is easiest to see across two conversations.

First:

```text
Customer:
The application crashes when I upload a PDF.

Agent:
Let's troubleshoot the PDF upload issue...
```

The interaction is retained.

Later:

```text
Customer:
My application is crashing again.
```

The agent recalls the customer's relevant history and can respond with something closer to:

```text
I remember you previously had an application crash while
uploading PDFs on Windows 11 with Chrome and application
version 2.4. Is the crash happening during another PDF upload?
```

The important part isn't the exact wording.

The important part is that the second interaction has context that wasn't present in the second message.

That context came from the memory layer.

## What I learned

### Memory should be a service boundary

The biggest architectural lesson was to resist putting memory logic directly into the agent.

The agent should orchestrate memory, not implement it.

That keeps retrieval, storage, customer isolation, and provider-specific behavior in one place.

### Retrieval and generation are different problems

I found it useful to ask two separate questions:

1. Did I retrieve the right information?
2. Did the model use that information correctly?

Those are different failure modes.

Treating them separately makes debugging much less ambiguous.

### Persistent memory should store outcomes

A support agent gets more useful over time when it remembers what worked and what didn't.

A transcript alone doesn't express that clearly enough.

### Async integration deserves its own boundary

The Hindsight client's asynchronous API was a good example of why infrastructure concerns should stay localized.

Once the adapter was in place, the rest of the application didn't need to care about event loops.

### Memory should be relevant, not merely available

The goal isn't to make the model aware of everything that ever happened.

The goal is to retrieve enough useful history to handle the current request better.

That distinction is what keeps persistent memory from becoming another form of context-window bloat.

## The part I would keep

If I had to reduce the architecture to one principle, it would be this:

```text
Don't make the agent remember everything.

Make the agent know where to ask for what matters.
```

That is where Hindsight fits in my system.

The LLM remains responsible for generating the response. Flask remains responsible for request orchestration. The frontend remains responsible for the customer interaction.

Hindsight sits underneath them as the persistent memory layer.

That separation gives me something I couldn't get from a larger prompt alone: a memory system I can inspect, isolate, debug, and evolve independently from the model.
